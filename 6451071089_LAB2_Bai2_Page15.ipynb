```python
print("Họ và tên: Nguyễn Công Vũ")
print("MSSV: 6451071089")
print("=" * 50)

import pandas as pd
import numpy as np
```

```python
train = pd.read_csv("/content/adult.data.csv")
test = pd.read_csv("/content/adult.test.csv")

print("Dữ liệu huấn luyện:")
display(train.head())

print("Dữ liệu kiểm thử:")
display(test.head())
```

```python
print("Số lượng dữ liệu huấn luyện:", len(train))
print("Số lượng dữ liệu kiểm thử:", len(test))
print("Tổng số lượng dữ liệu:", len(train) + len(test))
```

```python
# ==========================================
# CÂU 2: TIỀN XỬ LÝ DỮ LIỆU
# Sinh viên: Nguyễn Công Vũ
# MSSV: 6451071089
# ==========================================

# 1. Xóa các dòng chứa dữ liệu thiếu '?'
train = train.replace('?', np.nan).dropna().reset_index(drop=True)
test = test.replace('?', np.nan).dropna().reset_index(drop=True)

print("1. SAU KHI XÓA DỮ LIỆU THIẾU")
print("Số dòng Training:", len(train))
print("Số dòng Testing:", len(test))

# 2. Xóa cột fnlwgt
train = train.drop(columns=['fnlwgt'], errors='ignore')
test = test.drop(columns=['fnlwgt'], errors='ignore')

print("\n2. SAU KHI XÓA CỘT fnlwgt")
print("Số cột Training:", train.shape[1])
print("Số cột Testing:", test.shape[1])

# 3. Nối dữ liệu Training và Testing
data = pd.concat([train, test], ignore_index=True)

print("\n3. SAU KHI NỐI DỮ LIỆU")
print("Kích thước dữ liệu:", data.shape)
print("Số giá trị thiếu:", data.isnull().sum().sum())

display(data.head())
```

```python
# ==========================================
# CÂU 3: KHẢO SÁT TƯƠNG QUAN PEARSON
# ==========================================

# In thông tin sinh viên
print("Họ và tên: Nguyễn Công Vũ")
print("MSSV: 6451071089")
print("=" * 50)

import matplotlib.pyplot as plt
import seaborn as sns

# 1. Chọn các thuộc tính dạng số
numeric_data = data.select_dtypes(include=['number'])

# 2. Tính ma trận tương quan Pearson
corr_matrix = numeric_data.corr(method='pearson')

print("\nMA TRẬN TƯƠNG QUAN PEARSON:")
display(corr_matrix.round(3))

# 3. Vẽ biểu đồ Heatmap
plt.figure(figsize=(12, 8))
sns.heatmap(corr_matrix, annot=True, cmap='rocket_r', fmt='.3f')

plt.title("Pearson Correlation Heatmap - Adult Income")
plt.tight_layout()
plt.show()
```

```python
# ==========================================
# CÂU 4: TÁCH FEATURES VÀ LABELS
# ==========================================

print("Họ và tên: Nguyễn Công Vũ")
print("MSSV: 6451071089")
print("=" * 50)

# 1. Tách thuộc tính đầu vào
features = data.drop('income', axis=1)

# 2. Tách thuộc tính quyết định
labels = data['income']

# 3. Hiển thị kết quả
print("KÍCH THƯỚC FEATURES:", features.shape)
print("KÍCH THƯỚC LABELS:", labels.shape)

print("\nCÁC THUỘC TÍNH ĐẦU VÀO:")
print(features.columns.tolist())

print("\nTHUỘC TÍNH QUYẾT ĐỊNH:")
print(labels.name)

print("\n5 DÒNG FEATURES ĐẦU TIÊN:")
display(features.head())

print("\n5 DÒNG LABELS ĐẦU TIÊN:")
display(labels.head())
```

```python
# ==========================================
# CÂU 5: ONE-HOT ENCODING
# ==========================================

print("Họ và tên: Nguyễn Công Vũ")
print("MSSV: 6451071089")
print("=" * 50)

# 1. Xác định các thuộc tính không phải dạng số
categorical_cols = features.select_dtypes(exclude=['int64']).columns

print("CÁC THUỘC TÍNH KHÔNG PHẢI DẠNG SỐ:")
print(categorical_cols.tolist())
print("Số lượng thuộc tính:", len(categorical_cols))

# 2. Chuyển đổi sang One-Hot Encoding
features_onehot = pd.get_dummies(features, columns=categorical_cols)

# 3. Hiển thị kết quả
print("\nKÍCH THƯỚC DỮ LIỆU SAU KHI MÃ HÓA:")
print(features_onehot.shape)

print("\nDỮ LIỆU SAU KHI ONE-HOT ENCODING:")
display(features_onehot.head())
```

```python
# ==========================================
# CÂU 6: TÁCH DỮ LIỆU TRAINING VÀ TESTING
# ==========================================

print("Họ và tên: Nguyễn Công Vũ")
print("MSSV: 6451071089")
print("=" * 50)

# 1. Xác định số lượng dữ liệu ban đầu
n_train = 32561
n_test = len(features_onehot) - n_train

# 2. Tách dữ liệu huấn luyện và kiểm thử
X_train = features_onehot.iloc[:n_train]
X_test = features_onehot.iloc[n_train:]

y_train = labels.iloc[:n_train]
y_test = labels.iloc[n_train:]

# 3. Hiển thị kết quả
print("\nSỐ LƯỢNG DỮ LIỆU SAU KHI TÁCH:")
print("Training:", X_train.shape)
print("Testing:", X_test.shape)

print("\nSỐ LƯỢNG NHÃN:")
print("Training Labels:", y_train.shape)
print("Testing Labels:", y_test.shape)

print("\nTỔNG SỐ DỮ LIỆU:", len(X_train) + len(X_test))

print("\n5 DÒNG DỮ LIỆU TRAINING:")
display(X_train.head())

print("\n5 DÒNG DỮ LIỆU TESTING:")
display(X_test.head())
```
