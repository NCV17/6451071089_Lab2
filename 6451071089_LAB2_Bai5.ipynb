```python
# Họ và tên: Nguyễn Công Vũ
# MSSV: 6451071089

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Đọc dữ liệu
data = pd.read_csv('/content/heart.csv')

# 2. Đổi tên các thuộc tính
data.columns = [
    'age', 'sex', 'chest_pain_type',
    'resting_blood_pressure', 'cholesterol',
    'fasting_blood_sugar', 'rest_ecg',
    'max_heart_rate_achieved', 'exercise_induced_angina',
    'st_depression', 'st_slope',
    'num_major_vessels', 'thalassemia', 'target'
]

# 3. Xóa các dòng dữ liệu trống
data = data.dropna()

# 4. Hiển thị kết quả
print("Họ và tên: Nguyễn Công Vũ")
print("MSSV: 6451071089")
print("=" * 60)

print("\n1. KÍCH THƯỚC DỮ LIỆU SAU TIỀN XỬ LÝ:")
print(data.shape)

print("\n2. DANH SÁCH CÁC THUỘC TÍNH:")
print(data.columns.tolist())

print("\n3. KIỂM TRA GIÁ TRỊ THIẾU:")
print(data.isnull().sum())

print("\n4. 5 DÒNG DỮ LIỆU ĐẦU TIÊN:")
display(data.head())
```

### Câu 3. Khảo sát độ tương đồng giữa các thuộc tính bằng hệ số tương quan Pearson

```python
# Họ và tên: Nguyễn Công Vũ
# MSSV: 6451071089

print("Họ và tên: Nguyễn Công Vũ")
print("MSSV: 6451071089")
print("=" * 60)

# 1. Tính ma trận tương quan Pearson
correlation_matrix = data.corr(method='pearson')

print("\n1. MA TRẬN TƯƠNG QUAN PEARSON:")
display(correlation_matrix)

# 2. Trực quan hóa ma trận tương quan
plt.figure(figsize=(14, 10))
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title('Ma trận tương quan Pearson - Heart Disease')
plt.tight_layout()
plt.show()

# 3. Xác định các cặp thuộc tính có tương quan cao
threshold = 0.8

high_corr = []

for i in range(len(correlation_matrix.columns)):
    for j in range(i + 1, len(correlation_matrix.columns)):
        corr_value = correlation_matrix.iloc[i, j]

        if abs(corr_value) >= threshold:
            high_corr.append((
                correlation_matrix.columns[i],
                correlation_matrix.columns[j],
                corr_value
            ))

print("\n2. CÁC CẶP THUỘC TÍNH CÓ TƯƠNG QUAN CAO:")
if high_corr:
    for col1, col2, value in high_corr:
        print(f"{col1} - {col2}: {value:.4f}")
else:
    print("Không có cặp thuộc tính nào có tương quan từ 0.8 trở lên.")

# 4. Loại bỏ thuộc tính có tương quan cao
drop_columns = set()

for col1, col2, value in high_corr:
    if col1 != 'target' and col2 != 'target':
        drop_columns.add(col2)

data = data.drop(columns=list(drop_columns))

print("\n3. CÁC CỘT ĐÃ LOẠI BỎ:")
print(list(drop_columns))

print("\n4. CÁC THUỘC TÍNH SAU KHI LOẠI BỎ:")
print(data.columns.tolist())

print("\n5. KÍCH THƯỚC DỮ LIỆU SAU XỬ LÝ:")
print(data.shape)
```

```python
# ============================================================
# CÂU 4. TÁCH DỮ LIỆU THÀNH HAI PHẦN X VÀ y
# ============================================================

# 1. Tách thuộc tính đầu vào và thuộc tính quyết định
X = data.drop(columns=['target'])
y = data['target']

# 2. Hiển thị kết quả
print("Họ và tên: Nguyễn Công Vũ")
print("MSSV: 6451071089")
print("=" * 60)

print("\n1. KÍCH THƯỚC DỮ LIỆU:")
print("X:", X.shape)
print("y:", y.shape)

print("\n2. DANH SÁCH THUỘC TÍNH ĐẦU VÀO:")
print(X.columns.tolist())

print("\n3. THUỘC TÍNH QUYẾT ĐỊNH:")
print(y.name)

print("\n4. PHÂN BỐ GIÁ TRỊ TARGET:")
print(y.value_counts())

print("\n5. 5 DÒNG DỮ LIỆU ĐẦU TIÊN CỦA X:")
display(X.head())

print("\n6. 5 GIÁ TRỊ ĐẦU TIÊN CỦA y:")
display(y.head())
```

```python
# ============================================================
# CÂU 5. CHUYỂN ĐỔI THUỘC TÍNH SANG ONE-HOT ENCODING
# ============================================================

# 1. Chuyển đổi các thuộc tính phân loại bằng One-Hot Encoding
X_encoded = pd.get_dummies(
    X,
    columns=[
        'sex',
        'chest_pain_type',
        'fasting_blood_sugar',
        'rest_ecg',
        'exercise_induced_angina',
        'st_slope',
        'num_major_vessels',
        'thalassemia'
    ],
    dtype=int
)

# 2. Hiển thị kết quả
print("Họ và tên: Nguyễn Công Vũ")
print("MSSV: 6451071089")
print("=" * 60)

print("\n1. KÍCH THƯỚC DỮ LIỆU TRƯỚC KHI CHUYỂN ĐỔI:")
print(X.shape)

print("\n2. KÍCH THƯỚC DỮ LIỆU SAU KHI CHUYỂN ĐỔI:")
print(X_encoded.shape)

print("\n3. DANH SÁCH CÁC THUỘC TÍNH SAU KHI CHUYỂN ĐỔI:")
print(X_encoded.columns.tolist())

print("\n4. 5 DÒNG DỮ LIỆU SAU KHI CHUYỂN ĐỔI:")
display(X_encoded.head())

print("\n5. KIỂM TRA KIỂU DỮ LIỆU:")
print(X_encoded.dtypes.value_counts())
```

```python
# ============================================================
# CÂU 6. CHIA DỮ LIỆU HUẤN LUYỆN VÀ KIỂM THỬ
# ============================================================

from sklearn.model_selection import train_test_split

# 1. Chia dữ liệu theo tỷ lệ 80:20
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)

# 2. Hiển thị kết quả
print("Họ và tên: Nguyễn Công Vũ")
print("MSSV: 6451071089")
print("=" * 60)

print("\n1. KÍCH THƯỚC DỮ LIỆU BAN ĐẦU:")
print("X:", X_encoded.shape)
print("y:", y.shape)

print("\n2. KÍCH THƯỚC TẬP HUẤN LUYỆN:")
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("\n3. KÍCH THƯỚC TẬP KIỂM THỬ:")
print("X_test:", X_test.shape)
print("y_test:", y_test.shape)

print("\n4. PHÂN BỐ TARGET TRONG TẬP HUẤN LUYỆN:")
print(y_train.value_counts())

print("\n5. PHÂN BỐ TARGET TRONG TẬP KIỂM THỬ:")
print(y_test.value_counts())

print("\n6. HOÀN THÀNH CHIA DỮ LIỆU!")
```

```python
# ============================================================
# CÂU 7. XÂY DỰNG CÂY QUYẾT ĐỊNH ID3
# ============================================================

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# 1. Xây dựng mô hình ID3 sử dụng Entropy
clf_id3 = DecisionTreeClassifier(
    criterion='entropy',
    random_state=42
)

# 2. Huấn luyện mô hình
clf_id3.fit(X_train, y_train)

# 3. Dự đoán trên tập kiểm thử
y_pred_id3 = clf_id3.predict(X_test)

# 4. Đánh giá mô hình
print("Họ và tên: Nguyễn Công Vũ")
print("MSSV: 6451071089")
print("=" * 60)

print("\n1. KẾT QUẢ ĐÁNH GIÁ MÔ HÌNH ID3")

accuracy_id3 = accuracy_score(y_test, y_pred_id3)
print(f"Accuracy: {accuracy_id3:.4f}")

print("\n2. CLASSIFICATION REPORT:")
print(classification_report(y_test, y_pred_id3))

print("\n3. MA TRẬN NHẦM LẪN:")
cm_id3 = confusion_matrix(y_test, y_pred_id3)
print(cm_id3)

# 5. Biểu diễn ma trận nhầm lẫn
plt.figure(figsize=(6, 5))
sns.heatmap(
    cm_id3,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=['Không bệnh (0)', 'Có bệnh (1)'],
    yticklabels=['Không bệnh (0)', 'Có bệnh (1)']
)

plt.xlabel('Predicted label')
plt.ylabel('Actual label')
plt.title(f'Confusion Matrix - Decision Tree ID3 (Accuracy: {accuracy_id3:.4f})')
plt.tight_layout()
plt.show()

# 6. Biểu diễn cây quyết định
plt.figure(figsize=(24, 12))

plot_tree(
    clf_id3,
    feature_names=X_train.columns,
    class_names=['Không bệnh (0)', 'Có bệnh (1)'],
    filled=True,
    rounded=True,
    fontsize=8
)

plt.title('Decision Tree ID3 - Heart Disease')
plt.tight_layout()
plt.savefig('decision_tree_id3_heart.png', dpi=300, bbox_inches='tight')
plt.show()

print("\n4. ĐÃ LƯU CÂY QUYẾT ĐỊNH: decision_tree_id3_heart.png")
```

```python
# ============================================================
# CÂU 8. XÂY DỰNG CÂY QUYẾT ĐỊNH CART
# ============================================================

from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

print("Họ và tên: Nguyễn Công Vũ")
print("MSSV: 6451071089")
print("=" * 60)

# 1. Khởi tạo mô hình CART sử dụng Gini
clf_cart = DecisionTreeClassifier(
    criterion='gini',
    random_state=42
)

# 2. Huấn luyện mô hình
clf_cart.fit(X_train, y_train)

print("\n1. HUẤN LUYỆN MÔ HÌNH CART THÀNH CÔNG")

# 3. Dự đoán dữ liệu kiểm thử
y_pred_cart = clf_cart.predict(X_test)

# 4. Đánh giá mô hình
accuracy_cart = accuracy_score(y_test, y_pred_cart)

print("\n2. KẾT QUẢ ĐÁNH GIÁ MÔ HÌNH")
print("Accuracy:", accuracy_cart)

print("\nClassification Report:")
print(classification_report(y_test, y_pred_cart, zero_division=0))

# 5. Tính ma trận nhầm lẫn
cm_cart = confusion_matrix(y_test, y_pred_cart)

print("\n3. MA TRẬN NHẦM LẪN:")
print(cm_cart)

# 6. Biểu diễn ma trận nhầm lẫn
plt.figure(figsize=(7, 5))

sns.heatmap(
    cm_cart,
    annot=True,
    fmt='d',
    cmap='Greens',
    xticklabels=['Không bệnh (0)', 'Có bệnh (1)'],
    yticklabels=['Không bệnh (0)', 'Có bệnh (1)']
)

plt.xlabel('Predicted label')
plt.ylabel('Actual label')
plt.title(f'CART Confusion Matrix - Accuracy: {accuracy_cart:.4f}')
plt.tight_layout()
plt.show()

# 7. Biểu diễn cây quyết định
plt.figure(figsize=(24, 12))

plot_tree(
    clf_cart,
    feature_names=X_train.columns,
    class_names=['Không bệnh (0)', 'Có bệnh (1)'],
    filled=True,
    rounded=True,
    max_depth=3,
    fontsize=8
)

plt.title('Decision Tree CART - Heart Disease')
plt.tight_layout()
plt.savefig('decision_tree_cart_heart.png', dpi=150, bbox_inches='tight')
plt.show()

print("\n4. ĐÃ LƯU CÂY QUYẾT ĐỊNH: decision_tree_cart_heart.png")
print("\nHOÀN THÀNH CÂU 8!")
```

## CÂU 9. XÂY DỰNG MÔ HÌNH NAIVE BAYES VÀ ĐÁNH GIÁ KẾT QUẢ

```python
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

print("Họ và tên: Nguyễn Công Vũ")
print("MSSV: 6451071089")
print("=" * 60)

# 1. Xây dựng và huấn luyện mô hình Naive Bayes
clf_nb = GaussianNB()
clf_nb.fit(X_train, y_train)

# 2. Dự đoán trên tập kiểm tra
y_pred_nb = clf_nb.predict(X_test)

# 3. Đánh giá độ chính xác
accuracy_nb = accuracy_score(y_test, y_pred_nb)

print("ĐỘ CHÍNH XÁC NAIVE BAYES:", round(accuracy_nb * 100, 2), "%")

print("\nBÁO CÁO PHÂN LOẠI:")
print(classification_report(
    y_test,
    y_pred_nb,
    target_names=['Không bệnh (0)', 'Có bệnh (1)'],
    zero_division=0
))

# 4. Ma trận nhầm lẫn
cm_nb = confusion_matrix(y_test, y_pred_nb)

plt.figure(figsize=(6, 4))
sns.heatmap(
    cm_nb,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=['Không bệnh', 'Có bệnh'],
    yticklabels=['Không bệnh', 'Có bệnh']
)

plt.title('Confusion Matrix - Naive Bayes')
plt.xlabel('Nhãn dự đoán')
plt.ylabel('Nhãn thực tế')
plt.show()
```

## CÂU 10. SO SÁNH VÀ ĐÁNH GIÁ CÁC MÔ HÌNH

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("Họ và tên: Nguyễn Công Vũ")
print("MSSV: 6451071089")
print("=" * 60)

# 1. Tổng hợp kết quả các mô hình
results = pd.DataFrame({
    'Mô hình': ['ID3 (Entropy)', 'CART (Gini)', 'Naive Bayes'],
    'Accuracy (%)': [
        accuracy_id3 * 100,
        accuracy_cart * 100,
        accuracy_nb * 100
    ]
})

display(results.round(2))

# 2. Xác định mô hình có độ chính xác cao nhất
best_model = results.loc[results['Accuracy (%)'].idxmax()]

print("Mô hình có độ chính xác cao nhất:", best_model['Mô hình'])
print("Accuracy:", round(best_model['Accuracy (%)'], 2), "%")

# 3. Biểu đồ so sánh
plt.figure(figsize=(8, 5))
sns.barplot(
    data=results,
    x='Mô hình',
    y='Accuracy (%)'
)

plt.title('So sánh độ chính xác các mô hình')
plt.ylabel('Accuracy (%)')
plt.xlabel('Mô hình')
plt.ylim(0, 100)

for i, value in enumerate(results['Accuracy (%)']):
    plt.text(i, value + 1, f'{value:.2f}%', ha='center')

plt.show()
```

Qua kết quả thực nghiệm, mô hình Naive Bayes đạt độ chính xác cao nhất với 78,69%, tiếp theo là CART (Gini) với 73,77% và ID3 (Entropy) với 68,85%.

Như vậy, trong ba mô hình được thử nghiệm, Naive Bayes cho kết quả phân loại tốt nhất trên tập dữ liệu kiểm tra. Mô hình ID3 có độ chính xác thấp nhất, trong khi CART cho kết quả tốt hơn ID3.

**Kết luận:** Với bộ dữ liệu Heart Disease trong bài thực nghiệm này, Naive Bayes là mô hình có hiệu quả phân loại tốt nhất dựa trên chỉ số Accuracy.
