## 1. Mục tiêu
Tính chỉ số **F-measure (F1-score)** từ Precision và Recall để đánh giá, so sánh kết quả của các bài tập phân loại.
## 2. Công thức

$$
F1 = 2 \times \frac{Precision \times Recall}{Precision + Recall}
$$

F1-score là trung bình điều hòa của Precision và Recall. Giá trị càng gần 1 thì sự cân bằng giữa Precision và Recall càng tốt.

## 3. Xây dựng hàm tính F-measure

```python
print("Họ và tên: Nguyễn Công Vũ")
print("MSSV: 6451071089")
print("=" * 60)

def tinh_f_measure(precision, recall):
    if not (0 <= precision <= 1 and 0 <= recall <= 1):
        raise ValueError("Precision và Recall phải nằm trong khoảng từ 0 đến 1.")
    if precision + recall == 0:
        return 0.0
    return 2 * precision * recall / (precision + recall)

print("Đã xây dựng hàm tính F-measure thành công.")
```

## 4. Tính F-measure cho các trường hợp phân loại

Bảng dưới đây sử dụng các cặp Precision và Recall **minh họa** để kiểm tra cách tính. Khi có kết quả thực tế từ các bài tập trước, thay các giá trị này bằng số liệu tương ứng.

```python
print("Họ và tên: Nguyễn Công Vũ")
print("MSSV: 6451071089")
print("=" * 60)

# Dữ liệu minh họa (thay bằng kết quả thực tế nếu đã có)
cac_bai_tap = {
    "Trường hợp 1": {"precision": 0.80, "recall": 0.75},
    "Trường hợp 2": {"precision": 0.85, "recall": 0.80},
    "Trường hợp 3": {"precision": 0.90, "recall": 0.85}
}

print(f"{'Bài tập':<18}{'Precision':>12}{'Recall':>12}{'F-measure':>14}")
print("-" * 56)

for ten, chi_so in cac_bai_tap.items():
    p = chi_so["precision"]
    r = chi_so["recall"]
    f1 = tinh_f_measure(p, r)
    print(f"{ten:<18}{p:>12.4f}{r:>12.4f}{f1:>14.4f}")
```

## 5. Minh họa cách tính thủ công

**Với trường hợp 1:**

- Precision = 0.80
- Recall = 0.75

**Thay vào công thức:**

$$
F1 = 2 \times \frac{0.80 \times 0.75}{0.80 + 0.75}
$$

$$
F1 = \frac{1.20}{1.55} \approx 0.7742
$$

Vậy F-measure của trường hợp 1 xấp xỉ **0.7742 (77.42%)**.

## 6. Nhận xét và kết luận

- F-measure kết hợp Precision và Recall thành một chỉ số duy nhất.
- Khi Precision hoặc Recall thấp, F-measure thường giảm theo.
- Có thể dùng F-measure để so sánh các mô hình hoặc các bài tập phân loại trên cùng bộ dữ liệu.
- Trường hợp có F-measure cao hơn thể hiện sự cân bằng Precision và Recall tốt hơn.

**Kết luận:** F-measure là chỉ số hữu ích khi cần đánh giá đồng thời độ chính xác và độ phủ của mô hình phân loại.
